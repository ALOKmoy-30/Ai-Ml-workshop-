```python
# step 8: drop unnecessary columns
columns_to_drop = ["PassengerId", "Name", "Ticket", "Cabin"]
existing_drop_cols =[col for col in columns_to_drop if col in df.columns]
df.drop(columns=existing_drop_cols, inplace=True)
print("\n columns after droping unneccessary columns:")
print(df.columns)
```

```python
# step 9 :handle missing values
if "Age" in df.columns:
    df["Age"]= df["Age"].fillna(df["Age"].median())
if "Embarked" in df.columns:
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode([0]))
print("\nMissing values after cleaning")
print(df.isnull().sum())
```

```python
# step 10 : separate features (x) and target (y)
x = df.drop(columns=["Survived"])
y=df["Survived"]

print("\nInput features (x) preview:")
print(x.head())

print("\nTarget variable (y) preview:")
print( y.head())
```

```python
# step 11 : apply one-hot encoding to categorical input columns
categorical_columns = [col for col in["Sex","Embarked"] if col in x.columns ]
x = pd.get_dummies(x, columns=categorical_columns, drop_first=True)
print("\nInput features after one-hot encoding:")
print(x.head())
print("\nColumns after one-hot encoding:")
print(x.columns)
```

```python
# step 12: apply label encoding to the target column
label_encoder=LabelEncoder()
y_encoded= label_encoder.fit_transform(y)
print("\nEncoded target values:")
print( y_encoded[:10])
```

```python
# step 13: split data info training and testing sets
x_train,x_test,y_train,y_test, = train_test_split(x,y_encoded, test_size=0.2 , random_state=42)
print("\nTraining-Test split shapes")
print("x_train:", x_train.shape)
print("x_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
# step 14: apply standardization
standard_scaler = StandardScaler()
x_train_strandadized = standard_scaler.fit_transform(x_train)
x_test_strandadized = standard_scaler.transform(x_test)
print("\n first 5 rows after standardization")
print(x_train_strandadized[:5])
```

```python
# step 15: apply normalization
minmax_scaler = MinMaxScaler()
x_train_normalized = minmax_scaler.fit_transform(x_train)
x_test_normalized = minmax_scaler.transform(x_test)
print("\n first 5 rows after normalization")
print(x_train_normalized[:5])
```
